# Exercise Insurance

```python
# Probability a homeowner makes a claim
p_home_claim = 0.06

# Probability a car owner makes a liability claim
p_car_claim = 0.05

# Probability of NOT making a claim on each policy
p_no_home_claim = 1 - p_home_claim
p_no_car_claim = 1 - p_car_claim

# Assuming independence between the two events,
# the company profits only if there is no claim on EITHER policy
p_profit = p_no_home_claim * p_no_car_claim

print(f'Probability of no home claim: {p_no_home_claim}')
print(f'Probability of no car claim: {p_no_car_claim}')
print(f'Probability the company makes a profit on this customer: {p_profit:.4f}')
print(f'Probability the company makes a profit on this customer: {p_profit*100:.2f}%')
```

# Exercise applications (1)

```python
# Total number of applicants
total_applicants = 300

# Breakdown by education level
master_count = 60
bachelor_count = 90
none_count = 150

# Hiring rate for each group
p_hired_given_master = 0.30
p_hired_given_bachelor = 0.25
p_hired_given_none = 0.15

# Probability (proportion) of applicants in each education group
p_master = master_count / total_applicants
p_bachelor = bachelor_count / total_applicants
p_none = none_count / total_applicants

# Law of Total Probability:
# P(Hired) = P(Hired|Master)*P(Master) + P(Hired|Bachelor)*P(Bachelor) + P(Hired|None)*P(None)
p_hired = (p_hired_given_master * p_master +
           p_hired_given_bachelor * p_bachelor +
           p_hired_given_none * p_none)

print(f'P(Master degree)   = {p_master:.4f}')
print(f'P(Bachelor degree)  = {p_bachelor:.4f}')
print(f'P(No degree)        = {p_none:.4f}')
print()
print(f'Probability that an applicant is hired: {p_hired:.4f}')
print(f'Probability that an applicant is hired: {p_hired*100:.2f}%')
```
